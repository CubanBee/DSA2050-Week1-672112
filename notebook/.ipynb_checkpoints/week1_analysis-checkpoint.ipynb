```python
#Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Load the Dataset
df = pd.read_csv("../data/DSA_2050_Week1_Retail_Data (1).csv")
```

```python
#Inspect the Dataset
print(df.head())
print(df.shape)
print(df.info())
```

One row represents one single retail transaction.

```python
#Useful KPIs
#Total Revenue
total_revenue = df["Revenue_KES"].sum()
print("Total Revenue:", total_revenue)

#Units Sold
units_sold = df["Quantity"].sum()
print("Units Sold:", units_sold)

#Average Sales Value
avg_sales_value = total_revenue / units_sold
print("Average Sales Value:", avg_sales_value)
```

```python
#Visualization of Revenue by Region
df.groupby("Region")["Revenue_KES"].sum().plot(kind="bar", title="Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue (KES)")
plt.show()
```

Nairobi is the most profitable region since it generates the highest revenue.

```python
#Visualization of Units Sold by Category
df.groupby("Category")["Quantity"].sum().plot(kind="bar", title="Units Sold by Category")
plt.xlabel("Category")
plt.ylabel("Units Sold")
plt.show()
```

Electronics has the highest number of units sold. This shows that there is a strong demand in this category.

In conclusion, Nairobi and Electronics are leading in performance. The company should give first priority to the two, but more data is needed in order to understand customer demographics and long-term trends.
